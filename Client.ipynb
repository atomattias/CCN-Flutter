```python
# Install dependencies
!pip install requests

# Import libraries
import requests
from IPython.display import Image
from google.colab import files

# URL of the FastAPI server (replace with the ngrok URL from Notebook 1)
api_url = " https://80a1-34-86-128-50.ngrok-free.app/anonymize/"  # Add `/anonymize/` to the URL


# Upload a sample image to Colab
uploaded = files.upload()

# Get the path of the uploaded image
image_path = list(uploaded.keys())[0]
print("Uploaded image:", image_path)

# Send a POST request to the API
with open(image_path, "rb") as file:
    response = requests.post(api_url, files={"file": file})

# Check the response
if response.status_code == 200:
    # Save the anonymized image
    output_path = "/content/anonymized_image.jpg"
    with open(output_path, "wb") as f:
        f.write(response.content)
    print("Anonymized image saved to:", output_path)

    # Display the anonymized image
    display(Image(filename=output_path))
else:
    print(f"Error: {response.status_code}, {response.text}")
```
